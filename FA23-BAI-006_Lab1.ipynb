### Activity 1 — Clean a paragraph using core Python
Write a function that lowercases words and strips leading/trailing punctuation.

```python
import string

def clean_words(paragraph):
    words = paragraph.lower().split()
    cleaned = [word.strip(string.punctuation) for word in words]
    return [word for word in cleaned if word]

paragraph = "NLP is FUN! Python, strings, and tokens make text-processing easier."
print(clean_words(paragraph))
```

### Activity 2 — Ten most frequent words
Use `Counter` to print the ten most frequent words in an article as a table.

```python
from collections import Counter
import re
import pandas as pd

article = '''
Natural language processing helps computers process human language.
Language data needs cleaning before processing. Text processing includes
tokenization, cleaning, and counting words. Natural language tools process
text data and help people understand language.
'''

tokens = re.findall(r"\b[a-zA-Z]+\b", article.lower())
top_words = Counter(tokens).most_common(10)
frequency_table = pd.DataFrame(top_words, columns=["word", "count"])
display(frequency_table)
```

### Activity 3 — Sentence DataFrame
Create one row per sentence and add word and character counts with `.apply()`.

```python
sentences = [
    "Natural language processing is useful.",
    "Python can clean text.",
    "Pandas stores text data in tables."
]
df_sentences = pd.DataFrame({"sentence": sentences})
df_sentences["word_count"] = df_sentences["sentence"].apply(lambda s: len(s.split()))
df_sentences["char_count"] = df_sentences["sentence"].apply(len)
display(df_sentences)
```

### Activity 4 — Cosine similarity using NumPy
Compute cosine similarity for two word-count vector pairs and state which pair is more similar.

```python
import numpy as np

def cosine_similarity(a, b):
    a = np.array(a, dtype=float)
    b = np.array(b, dtype=float)
    denominator = np.linalg.norm(a) * np.linalg.norm(b)
    if denominator == 0:
        return 0.0
    return float(np.dot(a, b) / denominator)

# Pair A: word-count vectors with similar proportions
pair_a = cosine_similarity([3, 2, 1], [6, 4, 2])
# Pair B: word-count vectors with different proportions
pair_b = cosine_similarity([3, 2, 1], [1, 5, 4])

print(f"Pair A cosine similarity: {pair_a:.4f}")
print(f"Pair B cosine similarity: {pair_b:.4f}")
print("More similar:", "Pair A" if pair_a > pair_b else "Pair B")
print("Reason: cosine similarity compares vector direction, not vector length.")
```

### Activity 5 — Reviews JSON to DataFrame
Load a JSON file containing ten short reviews, add each row's most frequent word and word count, then sort by word count descending. A sample JSON file is created here so the notebook runs without an external file.

```python
import json
from collections import Counter
import re
import pandas as pd

sample_reviews = [
    "The app is easy and very useful",
    "Useful app but the app sometimes freezes",
    "Great design and great support",
    "The app is slow after the latest update",
    "Easy to use and easy to learn",
    "Support answered my question quickly",
    "The new update fixed the login problem",
    "Login problem still appears after update",
    "Very useful tools and useful reminders",
    "The design is clean and the app is fast"
]

with open("reviews.json", "w", encoding="utf-8") as f:
    json.dump(sample_reviews, f, ensure_ascii=False, indent=2)

with open("reviews.json", "r", encoding="utf-8") as f:
    reviews = json.load(f)

def top_word(review):
    words = re.findall(r"\b[a-zA-Z]+\b", review.lower())
    return Counter(words).most_common(1)[0][0] if words else ""

review_df = pd.DataFrame({"review": reviews})
review_df["word_count"] = review_df["review"].apply(lambda s: len(s.split()))
review_df["top_word"] = review_df["review"].apply(top_word)
review_df = review_df.sort_values("word_count", ascending=False).reset_index(drop=True)
display(review_df)
```
